```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
data = {
    "Student ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Student Name": [
        "Amit", "Neha", "Rahul", "Priya", "Rohan",
        "Kavya", "Arjun", "Simran", "Vikas", "Anjali"
    ],
    "Gender": [
        "Male", "Female", "Male", "Female", "Male",
        "Female", "Male", "Female", "Male", "Female"
    ],
    "Department": [
        "BCA", "BCA", "MCA", "BCA", "MCA",
        "MCA", "BCA", "MCA", "BCA", "MCA"
    ],
    "Attendance": [85, 92, 68, 78, 88, 95, 55, 73, 81, 90],
    "Study Hours": [4, 6, 3, 5, 7, 8, 2, 4, 6, 7],
    "Assignment Marks": [80, 90, 65, 75, 85, 92, 50, 70, 82, 88],
    "Internal Assessment Marks": [78, 88, 62, 74, 84, 90, 48, 68, 80, 86],
    "Python Marks": [82, 94, 60, 76, 87, 96, 45, 72, 85, 91],
    "DBMS Marks": [75, 89, 65, 70, 82, 93, 50, 68, 78, 88],
    "AI Marks": [80, 91, 58, 78, 85, 95, 48, 70, 81, 90],
    "Final Exam Marks": [84, 93, 61, 77, 86, 94, 52, 71, 83, 89]
}

df = pd.DataFrame(data)

print(df)
```

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns)

print("\nFirst 5 Students:")
print(df.head())
```

```python
# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())

# Check data types
print("\nData Types:")
print(df.dtypes)
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Handle missing values
df = df.dropna()

print("After Cleaning:")
print("Shape:", df.shape)
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())
```

```python
print(df.describe())
```

```python
df["Average Marks"] = df[
    [
        "Assignment Marks",
        "Internal Assessment Marks",
        "Python Marks",
        "DBMS Marks",
        "AI Marks",
        "Final Exam Marks"
    ]
].mean(axis=1)

print(df[["Student ID", "Student Name", "Average Marks"]])
```

```python
top_performers = df.sort_values(
    by="Average Marks",
    ascending=False
).head(3)

print("Top 3 Performers:")
print(
    top_performers[
        ["Student ID", "Student Name", "Department", "Average Marks"]
    ]
)
```

```python
low_performers = df.sort_values(
    by="Average Marks",
    ascending=True
).head(3)

print("Bottom 3 Performers:")
print(
    low_performers[
        ["Student ID", "Student Name", "Department", "Average Marks"]
    ]
)
```

```python
subject_columns = [
    "Python Marks",
    "DBMS Marks",
    "AI Marks"
]

print("Average Marks by Subject:")
print(df[subject_columns].mean())
```

```python
print("Maximum Marks by Subject:")
print(df[subject_columns].max())
```

```python
print("Minimum Marks by Subject:")
print(df[subject_columns].min())
```

```python
subject_summary = pd.DataFrame({
    "Average": df[subject_columns].mean(),
    "Maximum": df[subject_columns].max(),
    "Minimum": df[subject_columns].min()
})

print(subject_summary)
```

```python
department_average = df.groupby("Department")["Average Marks"].mean()

print("Department-wise Average Performance:")
print(department_average)
```

```python
department_subject = df.groupby("Department")[
    ["Python Marks", "DBMS Marks", "AI Marks"]
].mean()

print("Department-wise Subject Performance:")
print(department_subject)
```

```python
department_count = df["Department"].value_counts()

print("Students in Each Department:")
print(department_count)
```

```python
department_average.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Department-wise Average Performance")
plt.xlabel("Department")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.show()
```

```python
print("Attendance Statistics:")
print(df["Attendance"].describe())
```

```python
low_attendance = df[df["Attendance"] < 75]

print("Students with Attendance below 75%:")
print(
    low_attendance[
        ["Student ID", "Student Name", "Department", "Attendance", "Average Marks"]
    ]
)
```

```python
attendance_performance = df[
    ["Student Name", "Attendance", "Average Marks"]
].sort_values(
    by="Attendance",
    ascending=False
)

print(attendance_performance)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Attendance"],
    df["Average Marks"]
)

plt.title("Attendance vs Average Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Average Marks")

plt.show()
```

```python
correlation = df["Attendance"].corr(df["Average Marks"])

print("Correlation between Attendance and Average Marks:", correlation)
```

```python
study_performance = df[
    ["Student Name", "Study Hours", "Average Marks"]
].sort_values(
    by="Study Hours",
    ascending=False
)

print("Study Hours vs Average Marks:")
print(study_performance)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Study Hours"],
    df["Average Marks"]
)

plt.title("Study Hours vs Average Marks")
plt.xlabel("Study Hours")
plt.ylabel("Average Marks")

plt.show()
```

```python
study_correlation = df["Study Hours"].corr(
    df["Average Marks"]
)

print(
    "Correlation between Study Hours and Average Marks:",
    study_correlation
)
```

```python
relationship_analysis = df[
    ["Student Name", "Attendance", "Study Hours", "Average Marks"]
]

print(relationship_analysis)
```

```python
subject_average = df[
    ["Python Marks", "DBMS Marks", "AI Marks"]
].mean()

plt.figure(figsize=(8, 5))

subject_average.plot(kind="bar")

plt.title("Subject-wise Average Marks")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)

plt.show()
```

```python
top_5 = df.sort_values(
    by="Average Marks",
    ascending=False
).head(5)

plt.figure(figsize=(9, 5))

plt.bar(
    top_5["Student Name"],
    top_5["Average Marks"]
)

plt.title("Top 5 Students by Average Marks")
plt.xlabel("Student Name")
plt.ylabel("Average Marks")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["Attendance"],
    bins=5
)

plt.title("Attendance Distribution")
plt.xlabel("Attendance (%)")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["Study Hours"],
    bins=5
)

plt.title("Study Hours Distribution")
plt.xlabel("Study Hours")
plt.ylabel("Number of Students")

plt.show()
```

```python
print("========== STUDENT PERFORMANCE ANALYSIS SUMMARY ==========")

print("\nTotal Students:", len(df))

print("\nTop Performer:")
top_student = df.loc[df["Average Marks"].idxmax()]
print(top_student[["Student ID", "Student Name", "Department", "Average Marks"]])

print("\nLowest Performer:")
low_student = df.loc[df["Average Marks"].idxmin()]
print(low_student[["Student ID", "Student Name", "Department", "Average Marks"]])

print("\nDepartment-wise Average:")
print(df.groupby("Department")["Average Marks"].mean())

print("\nSubject-wise Average:")
print(df[["Python Marks", "DBMS Marks", "AI Marks"]].mean())

print("\nAttendance vs Performance Correlation:",
      df["Attendance"].corr(df["Average Marks"]))

print("\nStudy Hours vs Performance Correlation:",
      df["Study Hours"].corr(df["Average Marks"]))
```
