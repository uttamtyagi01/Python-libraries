```python
import cv2
import numpy as np

print("OpenCV Version:", cv2.__version__)
print("OpenCV Imported Successfully!")
```

```python
# Create a sample image
image = np.zeros((300, 400, 3), dtype=np.uint8)

image[:, :, 0] = np.linspace(0, 255, 400, dtype=np.uint8)
image[:, :, 1] = 150

cv2.imwrite("sample.jpg", image)

print("Sample Image Created Successfully!")
```

```python
import matplotlib.pyplot as plt

print("Matplotlib Imported Successfully!")
```

```python
image = cv2.imread("sample.jpg")

print("Image Shape:", image.shape)

plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
resized_image = cv2.resize(image, (200, 150))

print("Original Image Shape:", image.shape)
print("Resized Image Shape:", resized_image.shape)

plt.imshow(cv2.cvtColor(resized_image, cv2.COLOR_BGR2RGB))
plt.title("Resized Image")
plt.axis("off")
plt.show()
```

```python
height, width = image.shape[:2]

center = (width // 2, height // 2)

matrix = cv2.getRotationMatrix2D(center, 90, 1.0)

rotated_image = cv2.warpAffine(
    image,
    matrix,
    (width, height)
)

plt.imshow(cv2.cvtColor(rotated_image, cv2.COLOR_BGR2RGB))
plt.title("Rotated Image")
plt.axis("off")
plt.show()
```

```python
drawing = image.copy()

# Draw Line
cv2.line(drawing, (20, 20), (200, 20), (255, 0, 0), 3)

# Draw Rectangle
cv2.rectangle(drawing, (50, 50), (200, 150), (0, 255, 0), 3)

# Draw Circle
cv2.circle(drawing, (300, 100), 40, (0, 0, 255), 3)

# Add Text
cv2.putText(
    drawing, "OpenCV",
    (100, 250),
    cv2.FONT_HERSHEY_SIMPLEX,
    1, (255, 255, 255), 2
)

plt.imshow(cv2.cvtColor(drawing, cv2.COLOR_BGR2RGB))
plt.title("OpenCV Drawing Functions")
plt.axis("off")
plt.show()
```

```python
# Gaussian Blur
gaussian = cv2.GaussianBlur(image, (5, 5), 0)

# Median Blur
median = cv2.medianBlur(image, 5)

# Bilateral Filter
bilateral = cv2.bilateralFilter(image, 9, 75, 75)

plt.figure(figsize=(9, 6))

plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(cv2.cvtColor(gaussian, cv2.COLOR_BGR2RGB))
plt.title("Gaussian Blur")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(cv2.cvtColor(median, cv2.COLOR_BGR2RGB))
plt.title("Median Blur")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(cv2.cvtColor(bilateral, cv2.COLOR_BGR2RGB))
plt.title("Bilateral Filter")
plt.axis("off")

plt.tight_layout()
plt.show()
```
