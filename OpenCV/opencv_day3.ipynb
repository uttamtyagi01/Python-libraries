```python
import cv2
import numpy as np

canvas = np.zeros((400, 500, 3), dtype=np.uint8)

def mouse_event(event, x, y, flags, param):
    if event == cv2.EVENT_LBUTTONDOWN:
        cv2.circle(canvas, (x, y), 20, (0, 255, 0), -1)
        print("Mouse Clicked at:", x, y)

cv2.namedWindow("Mouse Events")
cv2.setMouseCallback("Mouse Events", mouse_event)

while True:
    cv2.imshow("Mouse Events", canvas)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cv2.destroyAllWindows()
```

```python
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades +
    "haarcascade_frontalface_default.xml"
)

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()

    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5
    )

    for (x, y, w, h) in faces:
        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

    cv2.imshow("Face Detection", frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import os
import cv2

face_detector = cv2.CascadeClassifier(
    cv2.data.haarcascades +
    "haarcascade_frontalface_default.xml"
)

recognizer = cv2.face.LBPHFaceRecognizer_create()

os.makedirs("face_dataset/person_1", exist_ok=True)
os.makedirs("face_dataset/person_2", exist_ok=True)

print("Face Recognition Model Created Successfully!")
```

```python
def capture_faces(label):
    cap = cv2.VideoCapture(0)
    count = 0

    os.makedirs(f"face_dataset/{label}", exist_ok=True)

    while count < 30:
        ret, frame = cap.read()

        if not ret:
            break

        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        faces = face_detector.detectMultiScale(gray, 1.1, 5)

        for (x, y, w, h) in faces[:1]:
            face = cv2.resize(gray[y:y+h, x:x+w], (200, 200))

            cv2.imwrite(
                f"face_dataset/{label}/{count}.jpg",
                face
            )

            count += 1

            cv2.rectangle(frame, (x, y), (x+w, y+h), (0,255,0), 2)

        cv2.imshow("Face Collection - Press Q to stop", frame)

        if cv2.waitKey(100) & 0xFF == ord("q"):
            break

    cap.release()
    cv2.destroyAllWindows()

    print(label, "Images Collected:", count)
```

```python
capture_faces("person_1")
```

```python
capture_faces("person_2")
```

```python
faces = []
labels = []

person_map = {
    0: "Person 1",
    1: "Person 2"
}

for label_id, folder in enumerate(["person_1", "person_2"]):
    folder_path = f"face_dataset/{folder}"

    for filename in os.listdir(folder_path):
        if filename.lower().endswith(".jpg"):
            path = os.path.join(folder_path, filename)

            face = cv2.imread(path, cv2.IMREAD_GRAYSCALE)

            if face is not None:
                face = cv2.resize(face, (200, 200))
                faces.append(face)
                labels.append(label_id)

recognizer.train(
    faces,
    np.array(labels, dtype=np.int32)
)

recognizer.save("face_recognizer.yml")

print("Face Recognition Model Trained Successfully!")
print("Total Images:", len(faces))
```

```python
cap = cv2.VideoCapture(0)

previous_frame = None

while True:
    ret, frame = cap.read()

    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    gray = cv2.GaussianBlur(gray, (21, 21), 0)

    if previous_frame is None:
        previous_frame = gray
        continue

    difference = cv2.absdiff(previous_frame, gray)

    _, thresh = cv2.threshold(
        difference, 25, 255, cv2.THRESH_BINARY
    )

    thresh = cv2.dilate(thresh, None, iterations=2)

    contours, _ = cv2.findContours(
        thresh,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    movement = False

    for contour in contours:
        if cv2.contourArea(contour) > 1500:
            movement = True

            x, y, w, h = cv2.boundingRect(contour)

            cv2.rectangle(
                frame, (x, y), (x+w, y+h),
                (0, 255, 0), 2
            )

    status = "Movement Detected" if movement else "No Movement"

    cv2.putText(
        frame, status, (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8, (0, 0, 255), 2
    )

    cv2.imshow("Human Activity - Motion Detection", frame)

    previous_frame = gray

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
