```python
# Pandas – Day 1

## Aim
To study Pandas library and understand Series and DataFrames
with practical Python examples.

## Introduction
Pandas is a Python library used for data manipulation and
data analysis.

It provides two main data structures:
1. Series – One-dimensional labeled data.
2. DataFrame – Two-dimensional tabular data.

## Applications
- Data Analysis
- Data Cleaning
- Data Manipulation
- Reading and Writing Data
- Statistical Analysis
```

```python
data = [10, 20, 30, 40, 50]
series = pd.Series(data)
print(series)
```

```python
## Pandas Series

A Series is a one-dimensional labeled array capable
of holding different types of data.

It contains:
- Data
- Index

Each element in a Series has an associated label.
```

```python
import pandas as pd
marks = [85, 90, 78, 95, 88]
student_marks = pd.Series(marks)
print(student_marks)
```

```python
marks = [85, 90, 78, 95]

students = ["Rahul", "Aman", "Vikas", "Rohit"]

series = pd.Series(marks, index=students)

print(series)
```

```python
student = {
    "Rahul": 85,
    "Aman": 90,
    "Vikas": 78,
    "Rohit": 95
}

series = pd.Series(student)

print(series)
```

```python
marks = pd.Series(
    [85, 90, 78, 95],
    index=["Rahul", "Aman", "Vikas", "Rohit"]
)

print("Rahul's Marks:", marks["Rahul"])

print("First Student Marks:", marks.iloc[0])

print("First Two Students:")
print(marks.iloc[0:2])
```

```python
marks = pd.Series([10, 20, 30, 40, 50])

print("Original Series:")
print(marks)

print("\nAddition:")
print(marks + 5)

print("\nMultiplication:")
print(marks * 2)

print("\nMaximum Value:", marks.max())

print("Minimum Value:", marks.min())

print("Average:", marks.mean())
```

```python
## Pandas DataFrame

A DataFrame is a two-dimensional data structure
containing rows and columns.

It is similar to an Excel spreadsheet or SQL table.

Features:
- Rows and columns
- Labeled indexes
- Different data types
- Data selection and manipulation
```

```python
import pandas as pd

data = {
    "Name": ["Rahul", "Aman", "Vikas", "Rohit"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 90, 78, 95]
}

df = pd.DataFrame(data)

print(df)
```

```python
students = [
    {"Name": "Rahul", "Age": 20, "City": "Delhi"},
    {"Name": "Aman", "Age": 21, "City": "Mumbai"},
    {"Name": "Vikas", "Age": 19, "City": "Dehradun"}
]

df = pd.DataFrame(students)

print(df)
```

```python
data = {
    "Name": ["Rahul", "Aman", "Vikas", "Rohit"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 90, 78, 95]
}

df = pd.DataFrame(data)

print("Shape:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nIndex:")
print(df.index)

print("\nData Types:")
print(df.dtypes)

print("\nFirst Two Rows:")
print(df.head(2))

print("\nLast Two Rows:")
print(df.tail(2))
```

```python
data = {
    "Name": ["Rahul", "Aman", "Vikas", "Rohit"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 90, 78, 95]
}

df = pd.DataFrame(data)

print("Names:")
print(df["Name"])

print("\nMarks:")
print(df["Marks"])

print("\nMultiple Columns:")
print(df[["Name", "Marks"]])
```

```python
data = {
    "Name": ["Rahul", "Aman", "Vikas", "Rohit"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 90, 78, 95]
}

df = pd.DataFrame(data)

print("First Row:")
print(df.iloc[0])

print("\nSecond Row:")
print(df.iloc[1])

print("\nRow with Index 2:")
print(df.loc[2])

print("\nFirst Two Rows:")
print(df.iloc[0:2])
```

```python
data = {
    "Name": ["Rahul", "Aman", "Vikas", "Rohit"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 90, 78, 95]
}

df = pd.DataFrame(data)

print("Students scoring more than 85 marks:")

result = df[df["Marks"] > 85]

print(result)
```

```python
data = {
    "Name": ["Rahul", "Aman", "Vikas"],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

df["City"] = ["Delhi", "Mumbai", "Dehradun"]

print(df)
```

```python
data = {
    "Name": ["Rahul", "Aman", "Vikas"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

# Updating values
df.loc[0, "Marks"] = 92

print("Updated DataFrame:")
print(df)

# Removing Age column
df = df.drop(columns=["Age"])

print("\nAfter Removing Age Column:")
print(df)
```

```python
data = {
    "Name": ["Rahul", "Aman", "Vikas", "Rohit"],
    "Marks": [85, 90, 78, 95]
}

df = pd.DataFrame(data)

print("Total Marks:", df["Marks"].sum())

print("Average Marks:", df["Marks"].mean())

print("Highest Marks:", df["Marks"].max())

print("Lowest Marks:", df["Marks"].min())

print("\nStatistical Summary:")
print(df.describe())
```

```python
## Difference Between Series and DataFrame

| Series | DataFrame |
|---|---|
| One-dimensional | Two-dimensional |
| Single labeled column | Multiple labeled columns |
| Contains one axis | Contains two axes |
| Similar to a single column | Similar to an Excel table |
```

```python
## Practical Exercise

Create a student dataset using Pandas DataFrame.

Perform the following operations:
1. Display all student records.
2. Display students scoring above 80.
3. Calculate average marks.
4. Find the highest marks.
5. Display only Name and Marks columns.
```

```python
import pandas as pd

data = {
    "Name": ["Amit", "Rahul", "Priya", "Neha", "Karan"],
    "Age": [20, 21, 20, 19, 22],
    "Marks": [75, 88, 92, 68, 85]
}

df = pd.DataFrame(data)

print("Student Records:")
print(df)

print("\nStudents scoring above 80:")
print(df[df["Marks"] > 80])

print("\nAverage Marks:")
print(df["Marks"].mean())

print("\nHighest Marks:")
print(df["Marks"].max())

print("\nName and Marks:")
print(df[["Name", "Marks"]])
```
