```python
# Pandas – Day 2

## Aim
To learn how to read and write different types of
files using the Pandas library.

## Topics Covered
1. Reading CSV Files
2. Reading Excel Files
3. Reading JSON Files
4. Reading Text Files
5. Reading TSV Files
6. Important Parameters of read_csv()
```

```python
import pandas as pd
import os

print("Pandas Version:", pd.__version__)
print("Libraries Imported Successfully!")
```

```python
data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Rahul", "Aman", "Priya", "Neha", "Vikas"],
    "Age": [20, 21, 19, 22, 20],
    "Course": ["BCA", "BCA", "BBA", "BCA", "BBA"],
    "Marks": [85, 90, 78, 95, 88],
    "City": ["Delhi", "Mumbai", "Dehradun", "Pune", "Jaipur"]
}

df = pd.DataFrame(data)

print(df)
```

```python
os.makedirs("datasets", exist_ok=True)

print("Dataset folder created successfully!")
```

```python
df.to_csv("datasets/students.csv", index=False)

print("CSV file created successfully!")
```

```python
## Reading Excel Files

Excel files are used to store data in spreadsheet format.

Pandas provides:
1. to_excel() – To save DataFrame into Excel file.
2. read_excel() – To read Excel file into DataFrame.
```

```python
%pip install openpyxl
```

```python
df.to_excel("datasets/students.xlsx", index=False)

print("Excel file created successfully!")
```

```python
excel_data = pd.read_excel("datasets/students.xlsx")

print(excel_data)
```

```python
excel_data = pd.read_excel(
    "datasets/students.xlsx",
    sheet_name=0
)

print(excel_data)
```

```python
## Reading JSON Files

JSON stands for JavaScript Object Notation.

It is a lightweight data format used to store and
exchange structured information.

Pandas provides:
1. to_json() – To save DataFrame as JSON.
2. read_json() – To read JSON files.
```

```python
df.to_json(
    "datasets/students.json",
    orient="records",
    indent=4
)

print("JSON file created successfully!")
```

```python
json_data = pd.read_json("datasets/students.json")

print(json_data)
```

```python
print("First Five Records:")
print(json_data.head())

print("\nDataset Shape:")
print(json_data.shape)

print("\nColumn Names:")
print(json_data.columns)
```

```python
## Reading Text Files

Text files can contain structured data separated by
different delimiters.

Pandas read_csv() can also read text files by
specifying the appropriate separator.
```

```python
text_data = """Name|Age|Marks
Rahul|20|85
Aman|21|90
Priya|19|78
Neha|22|95
"""

with open("datasets/students.txt", "w") as file:
    file.write(text_data)

print("Text file created successfully!")
```

```python
text_df = pd.read_csv(
    "datasets/students.txt",
    sep="|"
)

print(text_df)
```

```python
## Reading TSV Files

TSV stands for Tab Separated Values.

In TSV files, values are separated using tab characters.

Pandas can read TSV files using read_csv()
with sep="\\t".
```

```python
df.to_csv(
    "datasets/students.tsv",
    sep="\t",
    index=False
)

print("TSV file created successfully!")
```

```python
tsv_data = pd.read_csv(
    "datasets/students.tsv",
    sep="\t"
)

print(tsv_data)
```

```python
data = pd.read_csv(
    "datasets/students.csv",
    usecols=["Name", "Marks"]
)

print(data)
```

```python
data = pd.read_csv(
    "datasets/students.csv",
    nrows=3
)

print(data)
```
