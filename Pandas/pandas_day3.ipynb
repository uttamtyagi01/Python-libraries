```python
# Pandas – Day 3: Data Cleaning

## Aim
To study and perform different data cleaning
operations using the Pandas library.

## Introduction
Data Cleaning is the process of identifying and
correcting errors, missing values, incorrect formats
and duplicate records in a dataset.

## Topics Covered
1. Cleaning Empty Cells
2. Cleaning Wrong Format
3. Cleaning Wrong Data
4. Removing Duplicates
```

```python
import pandas as pd
import numpy as np

print("Pandas Version:", pd.__version__)
print("Libraries Imported Successfully!")
```

```python
data = {
    "Student_ID": [101, 102, 103, 104, 105, 105, 106],
    "Name": ["Rahul", "Aman", "Priya", "Neha", "Vikas", "Vikas", "Arjun"],
    "Age": ["20", "21", None, "twenty-two", "-5", "-5", "23"],
    "Course": ["BCA", "BCA", "BBA", "bca", "BCA", "BCA", "BBA"],
    "Marks": [85, None, 78, 105, 88, 88, 92],
    "Join_Date": [
        "2026-01-10", "11/02/2026", "2026-03-15",
        "not available", "2026-04-20", "2026-04-20",
        "2026-05-01"
    ],
    "City": ["Delhi", "Mumbai", None, "Pune", "Jaipur", "Jaipur", "Dehradun"]
}

df = pd.DataFrame(data)

print(df)
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe(include="all"))
```

```python
print("Missing Values:")
print(df.isnull())

print("\nTotal Missing Values:")
print(df.isnull().sum())
```

```python
cleaned_df = df.dropna()

print(cleaned_df)
```

```python
cleaned_df = df.dropna(axis=1)

print(cleaned_df)
```

```python
cleaned_df = df.copy()

cleaned_df["Age"] = cleaned_df["Age"].fillna("Not Available")

cleaned_df["Marks"] = cleaned_df["Marks"].fillna(
    cleaned_df["Marks"].mean()
)

cleaned_df["City"] = cleaned_df["City"].fillna("Unknown")

print(cleaned_df)
```

```python
cleaned_df = df.copy()

cleaned_df = cleaned_df.fillna({
    "Age": "Unknown",
    "Marks": 0,
    "City": "Not Provided"
})

print(cleaned_df)
```

```python
cleaned_df = df.copy()

cleaned_df["Age"] = pd.to_numeric(
    cleaned_df["Age"],
    errors="coerce"
)

print(cleaned_df["Age"])
```

```python
cleaned_df["Join_Date"] = pd.to_datetime(
    cleaned_df["Join_Date"],
    errors="coerce",
    dayfirst=True
)

print(cleaned_df["Join_Date"])
```

```python
print("Updated Data Types:")
print(cleaned_df.dtypes)

print("\nCleaned Dataset:")
print(cleaned_df)
```

```python
wrong_data = df.copy()

wrong_data["Age"] = pd.to_numeric(
    wrong_data["Age"],
    errors="coerce"
)

print("Students with Negative Age:")

print(wrong_data[wrong_data["Age"] < 0])
```

```python
valid_age = wrong_data.loc[
    wrong_data["Age"] >= 0, "Age"
]

average_age = valid_age.median()

wrong_data.loc[
    wrong_data["Age"] < 0, "Age"
] = average_age

print(wrong_data[["Name", "Age"]])
```

```python
print("Marks Greater Than 100:")

print(wrong_data[wrong_data["Marks"] > 100])
```

```python
valid_marks = wrong_data.loc[
    wrong_data["Marks"].between(0, 100), "Marks"
]

average_marks = valid_marks.mean()

wrong_data.loc[
    (wrong_data["Marks"] < 0) |
    (wrong_data["Marks"] > 100),
    "Marks"
] = average_marks

print(wrong_data[["Name", "Marks"]])
```

```python
wrong_data["Course"] = wrong_data["Course"].str.upper()

print(wrong_data[["Name", "Course"]])
```

```python
cleaned_df = df.drop_duplicates(
    subset=["Student_ID"],
    keep="first"
)

print(cleaned_df)
```

```python
cleaned_df = df.copy()

# 1. Convert Age into numeric
cleaned_df["Age"] = pd.to_numeric(
    cleaned_df["Age"],
    errors="coerce"
)

# 2. Convert Marks into numeric
cleaned_df["Marks"] = pd.to_numeric(
    cleaned_df["Marks"],
    errors="coerce"
)

# 3. Correct invalid Age
cleaned_df.loc[
    cleaned_df["Age"] < 0, "Age"
] = np.nan

# 4. Correct invalid Marks
cleaned_df.loc[
    ~cleaned_df["Marks"].between(0, 100), "Marks"
] = np.nan

# 5. Fill missing Age and Marks
cleaned_df["Age"] = cleaned_df["Age"].fillna(
    cleaned_df["Age"].median()
)

cleaned_df["Marks"] = cleaned_df["Marks"].fillna(
    cleaned_df["Marks"].mean()
)

# 6. Standardize Course names
cleaned_df["Course"] = cleaned_df["Course"].str.upper()

# 7. Fill missing City
cleaned_df["City"] = cleaned_df["City"].fillna("Unknown")

# 8. Convert Date format
cleaned_df["Join_Date"] = pd.to_datetime(
    cleaned_df["Join_Date"],
    errors="coerce",
    dayfirst=True
)

# 9. Remove duplicate Student IDs
cleaned_df = cleaned_df.drop_duplicates(
    subset=["Student_ID"],
    keep="first"
)

print("Final Cleaned Dataset:")
print(cleaned_df)
```

```python
print("Missing Values:")
print(cleaned_df.isnull().sum())

print("\nDuplicate Records:")
print(cleaned_df.duplicated().sum())

print("\nFinal Dataset Shape:")
print(cleaned_df.shape)
```
