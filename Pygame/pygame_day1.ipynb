```python
import pygame

pygame.init()

screen = pygame.display.set_mode((600, 400))
pygame.display.set_caption("My First Pygame Window")

running = True

while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

        if event.type == pygame.KEYDOWN:
            if event.key == pygame.K_ESCAPE:
                running = False

    screen.fill((30, 30, 30))
    pygame.display.flip()

pygame.quit()

print("Game Window Closed!")
```

```python
import pygame

pygame.init()

screen = pygame.display.set_mode((600, 400))
pygame.display.set_caption("Pygame Shapes")

running = True

while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    screen.fill((255, 255, 255))

    # Rectangle
    pygame.draw.rect(screen, (255, 0, 0), (50, 50, 150, 100))

    # Circle
    pygame.draw.circle(screen, (0, 0, 255), (350, 100), 50)

    # Line
    pygame.draw.line(screen, (0, 150, 0), (50, 250), (500, 250), 5)

    pygame.display.flip()

pygame.quit()
```

```python
import pygame

pygame.init()

screen = pygame.display.set_mode((600, 400))
pygame.display.set_caption("Pygame Text")

font = pygame.font.Font(None, 50)

text = font.render("Hello Pygame!", True, (255, 255, 255))

running = True

while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    screen.fill((30, 30, 30))

    screen.blit(text, (150, 170))

    pygame.display.flip()

pygame.quit()
```
