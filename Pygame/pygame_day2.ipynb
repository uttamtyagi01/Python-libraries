```python
import pygame

pygame.init()

print("Pygame Version:", pygame.version.ver)
print("Pygame Day 2 Started!")
```

```python
screen = pygame.display.set_mode((600, 400))
pygame.display.set_caption("Keyboard and Mouse Events")

circles = []
running = True

while running:
    for event in pygame.event.get():

        if event.type == pygame.QUIT:
            running = False

        if event.type == pygame.MOUSEBUTTONDOWN:
            circles.append(event.pos)

        if event.type == pygame.KEYDOWN:
            if event.key == pygame.K_c:
                circles.clear()

            if event.key == pygame.K_ESCAPE:
                running = False

    screen.fill((30, 30, 30))

    for position in circles:
        pygame.draw.circle(screen, (0, 255, 0), position, 20)

    pygame.display.flip()

pygame.quit()
```

```python
import pygame

pygame.init()

screen = pygame.display.set_mode((600, 400))
pygame.display.set_caption("Object Movement")

x, y = 275, 175
speed = 5

clock = pygame.time.Clock()
running = True

while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    keys = pygame.key.get_pressed()

    if keys[pygame.K_LEFT]:
        x -= speed
    if keys[pygame.K_RIGHT]:
        x += speed
    if keys[pygame.K_UP]:
        y -= speed
    if keys[pygame.K_DOWN]:
        y += speed

    x = max(0, min(x, 560))
    y = max(0, min(y, 360))

    screen.fill((30, 30, 30))

    pygame.draw.rect(
        screen, (0, 255, 0), (x, y, 40, 40)
    )

    pygame.display.flip()
    clock.tick(60)

pygame.quit()
```

```python
import pygame

pygame.init()

screen = pygame.display.set_mode((600, 400))
pygame.display.set_caption("Collision Detection")

player = pygame.Rect(50, 150, 40, 40)
target = pygame.Rect(400, 150, 60, 60)

speed = 5
clock = pygame.time.Clock()
font = pygame.font.Font(None, 35)

running = True

while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    keys = pygame.key.get_pressed()

    if keys[pygame.K_LEFT]:
        player.x -= speed
    if keys[pygame.K_RIGHT]:
        player.x += speed
    if keys[pygame.K_UP]:
        player.y -= speed
    if keys[pygame.K_DOWN]:
        player.y += speed

    player.clamp_ip(screen.get_rect())

    collision = player.colliderect(target)

    screen.fill((30, 30, 30))

    pygame.draw.rect(screen, (0, 255, 0), player)

    target_color = (255, 0, 0) if not collision else (255, 255, 0)
    pygame.draw.rect(screen, target_color, target)

    message = "Collision Detected!" if collision else "Move the Green Box"
    text = font.render(message, True, (255, 255, 255))
    screen.blit(text, (150, 30))

    pygame.display.flip()
    clock.tick(60)

pygame.quit()
```
