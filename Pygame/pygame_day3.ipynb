```python
import pygame
import random

pygame.init()

print("Pygame Initialized Successfully!")
```

```python
screen = pygame.display.set_mode((600, 400))
pygame.display.set_caption("Catch the Falling Ball")

clock = pygame.time.Clock()
font = pygame.font.Font(None, 35)

player = pygame.Rect(250, 350, 100, 20)
ball = pygame.Rect(random.randint(0, 580), 0, 20, 20)

score = 0
lives = 3
speed = 4
running = True
game_over = False

while running:

    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

        if event.type == pygame.KEYDOWN:
            if event.key == pygame.K_ESCAPE:
                running = False

    if not game_over:

        keys = pygame.key.get_pressed()

        if keys[pygame.K_LEFT]:
            player.x -= 6

        if keys[pygame.K_RIGHT]:
            player.x += 6

        player.clamp_ip(screen.get_rect())

        ball.y += speed

        if player.colliderect(ball):
            score += 1
            ball.x = random.randint(0, 580)
            ball.y = 0
            speed = min(speed + 0.2, 10)

        elif ball.y > 400:
            lives -= 1
            ball.x = random.randint(0, 580)
            ball.y = 0

            if lives == 0:
                game_over = True

    screen.fill((30, 30, 30))

    pygame.draw.rect(screen, (0, 255, 0), player)
    pygame.draw.ellipse(screen, (255, 0, 0), ball)

    screen.blit(font.render(f"Score: {score}", True, (255,255,255)), (20, 20))
    screen.blit(font.render(f"Lives: {lives}", True, (255,255,255)), (450, 20))

    if game_over:
        message = font.render("GAME OVER!", True, (255, 255, 0))
        screen.blit(message, (220, 180))

    pygame.display.flip()
    clock.tick(60)

pygame.quit()
print("Game Finished! Final Score:", score)
```
