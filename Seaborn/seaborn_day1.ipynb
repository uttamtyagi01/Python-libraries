```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

print("Seaborn Version:", sns.__version__)
print("Libraries Imported Successfully!")
```

```python
data = {
    "Student": ["Rahul", "Aman", "Priya", "Neha", "Vikas",
                "Rohit", "Karan", "Anjali"],
    "Course": ["BCA", "BCA", "BBA", "BBA",
               "BCA", "BBA", "BCA", "BBA"],
    "Marks": [85, 90, 78, 95, 88, 82, 75, 92]
}

df = pd.DataFrame(data)

print(df)
```

```python
sns.boxplot(data=df, x="Course", y="Marks")

plt.title("Student Marks Distribution")
plt.xlabel("Course")
plt.ylabel("Marks")

plt.show()
```

```python
sns.violinplot(data=df, x="Course", y="Marks", palette="Set2")

plt.title("Student Marks Distribution using Violin Plot")
plt.xlabel("Course")
plt.ylabel("Marks")

plt.show()
```

```python
marks_data = pd.DataFrame({
    "Maths": [85, 90, 78, 95, 88],
    "Python": [90, 85, 80, 92, 87],
    "DBMS": [80, 88, 75, 90, 85]
})

correlation = marks_data.corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Student Marks Correlation Heatmap")
plt.show()
```

```python
sns.set_theme(style="darkgrid", palette="pastel")

sns.barplot(data=df, x="Student", y="Marks", hue="Course")

plt.title("Student Performance")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.xticks(rotation=45)

plt.show()
```
