```python
import numpy as np
```

```python
# np.array()
# 1d array
numbers=np.array([10,20,30,40])
```

```python
result=numbers+5
print(result)
```

```python
# 2d array
b=np.array([[1,2,3],
            [4,5,6]])
print(b)
```

```python
# np.zeros()
a=np.zeros(5)
print(a)
```

```python
b=np.zeros((2,3))
print(b)
```

```python
# np.ones()
a=np.ones(3)
print(a)
```

```python
b=np.ones((2,3))
print(b)
```

```python
# np.arange()
c=np.arange(1,9)
print(c)
```

```python
# np.linspace()
a=np.linspace(1,10,5)
print(a)
```

```python
# np.full()
a=np.full((2,3),6)
print(a)
```

```python
# np.eye()
a=np.eye(2)
print(a)
```

```python
a = np.array([[10, 20, 30],
              [40, 50, 60]])

print("Dimensions:", a.ndim)
print("Shape:", a.shape)
print("Size:", a.size)
print("Data type:", a.dtype)
```
