```python
import numpy as np
```

```python
# ARITHMETIC OPERATIONS
a=np.array([10,20,20])
b=np.array([30,40,50])
```

```python
# addition
print(a+b)
```

```python
# subtraction
print(b-a)
```

```python
# multiplication
print(a*b)
```

```python
# division
print(a/b)
```

```python
# power
print(a**2)
```

```python
# Remainder
print(a%3)
```

```python
# SCALER OPERATIONS
A = np.array([10, 20, 30])

print(A + 5)
print(A * 2)
print(A / 10)
```

```python
# MATHEMATICAL FUNCTIONS
A = np.array([1, 4, 9, 16])
print(np.sqrt(A))
```

```python
A = np.array([1, 2, 3, 4])
print(np.power(A, 3))
```

```python
# COMPARISON OPERATIONS
A = np.array([10, 20, 30])

print(A > 15)
print(A < 30)
print(A == 20)
print(A != 20)
```

```python
# LOGICAL OPERATIONS
A = np.array([10, 20, 30, 40, 50])

mask = (A >= 20) & (A <= 40)
print(mask)
print(A[mask])
```

```python
# AGGREGATE FUNCTIONS
marks = np.array([78, 90, 85, 65, 92])

print("Total:", np.sum(marks))
print("Average:", np.mean(marks))
print("Highest:", np.max(marks))
print("Lowest:", np.min(marks))
```
