```python
import numpy as np
marks = np.array([75, 80, 92, 68, 85])
print(marks)
```

```python
# Array save
import numpy as np
marks = np.array([75, 80, 92, 68, 85])
np.save("marks.npy", marks)
print("Array saved successfully!")
```

```python
loaded_marks = np.load("marks.npy")
print(loaded_marks)
```

```python
data = np.array([
    [10, 20],
    [30, 40],
    [50, 60]
])
np.savetxt("data.csv", data, delimiter=",", fmt="%d")
```

```python
loaded_data = np.loadtxt(
    "data.csv",
    delimiter=",",
    dtype=int
)
print(loaded_data)
```

```python
# RANDOM MODULE
import numpy as np
print(np.random.randint(1, 10))
```

```python
print(np.random.rand(3))
```

```python
values = np.array([10, 20, 30, 40])
print(np.random.choice(values, size=3))
```

```python
arr = np.array([1, 2, 3, 4, 5])
np.random.shuffle(arr)
print(arr)
```

```python
arr = np.array([1, 2, 3, 4, 5])
new_arr = np.random.permutation(arr)
print("Original:", arr)
print("New:", new_arr)
```

```python
# NORMAL DISTRIBUTION
values = np.random.normal(loc=50, scale=10, size=5)
print(values)
```

```python
values = np.random.binomial(n=10, p=0.5, size=5)
print(values)
```
